## Connector 객체
- db 연결 정보를 작성 -> 작성된 정보를 통해서 실제로 DB 서버와 연결하는 객체

## Cursor
- SQL을 수행하고 결과를 반환 받는 객체

**두 객체 모두 사용 후 `close()` 필수**

```python
import mysql.connector

# DB 연결 정보
config = {
    'host' : 'localhost', # 서버 주소
    'port' : 3306, # 서버 내 프로그램 접속 번호 (기본 값 3306 - 승낙 가능)
    'user' : 'skn_ai', #
    'password' : '1234',
    'database' : 'menudb'
}

# Connector 객체 생성
conn = mysql.connector.connect(**config) # ** == unpacking

# cursor 객체 생성
cursor = conn.cursor()

# select 수행
# 주의 !! SQL 마지막에 ; 작성 금지
cursor.execute('select * from tbl_menu')

rows = cursor.fetchall()

for row in rows:
    print(row)

# close는 역순 필수
cursor.close()
conn.close()
```

```python
class Menu:
    def __init__(self,
                 menu_code, menu_name, menu_price,
                 category_code, orderable_status):
        self.__menu_code = menu_code
        self.__menu_name = menu_name
        self.__menu_price = menu_price
        self.__category_code = category_code
        self.__orderable_status = orderable_status

    @property
    def menu_code(self):
        return self.__menu_code

    @property
    def menu_name(self):
        return self.__menu_name

    @property
    def menu_price(self):
        return self.__menu_price

    @property
    def category_code(self):
        return self.__category_code

    @property
    def orderable_status(self):
        return self.__orderable_status

    def __repr__(self):
        return (
            f"Menu(menu_code={self.__menu_code}, "
            f"menu_name={self.__menu_name}, "
            f"menu_price={self.__menu_price}, "
            f"category_code={self.__category_code}, "
            f"orderable_status={self.__orderable_status})"
        )
```

```python
# 메뉴 테이블 전체 조회해서 Menu 리스트에 담기
try:
    with mysql.connector.connect(**config) as conn: # DB 연결 담당
        with conn.cursor() as cursor: # SQL 수행 담당

            cursor.execute('select * from tbl_menu') # 실행

            rows = cursor.fetchall() # 조회 결과 집합 반환

            # 조회 결과를 Menu 객체 List로 변환
            menus = [Menu(*row) for row in rows]

            # menus = []
            # for row in rows:
            #     menus.append(Menu(*row))

    for menu in menus:
        print(menu)

except mysql.connector.Error as err:
    print("DB 오류 : ", err)
```

```python
# 밥 만 조회
input_menu_name = input('포함 메뉴명')

try:
    with mysql.connector.connect(**config) as conn: # DB 연결 담당
        with conn.cursor() as cursor: # SQL 수행 담당

            cursor.execute(f'''
                           select * from tbl_menu
                           where menu_name like '%{input_menu_name}%'
                           ''') # 실행

            rows = cursor.fetchall() # 조회 결과 집합 반환

            # 조회 결과를 Menu 객체 List로 변환
            menus = [Menu(*row) for row in rows]

            # menus = []
            # for row in rows:
            #     menus.append(Menu(*row))
    for menu in menus:
        print(menu)

except mysql.connector.Error as err:
    print("DB 오류 : ", err)
```

```python
new_menu = Menu(None, '멸치액젓맛요거트', 20000, 8, 'Y')

try:
    with mysql.connector.connect(**config) as conn:
        with conn.cursor() as cursor:
            cursor.execute('''
                insert into tbl_menu(
                    menu_name,
                    menu_price,
                    category_code,
                    orderable_status)
                values(%s, %s, %s, %s)
            ''', (#  mysql connectort에서 %s == 빈칸 하나
                new_menu.menu_name, # %s 자리 대입 시 자동으로 양쪽에 '' 작성
                new_menu.menu_price,
                new_menu.category_code,
                new_menu.orderable_status
            ))

            # 반영된 행의 수 확인
            print("insert된 행의 수: ", cursor.rowcount)

            # 커밋
            conn.commit()

except mysql.connector.Error as err:
    print("DB 에러:", err)
```

## 레코드 수정 (Update)

```python
# 추가된 메뉴의 가격을 10000 인하
target_menu_code = 104

try:
    with mysql.connector.connect(**config) as conn:
        with conn.cursor() as cursor:

            cursor.execute('''
                update tbl_menu
                set menu_price = menu_price - 10000
                where menu_code = %s
            ''', (target_menu_code, ))# 튜플로 입력해야함

            print('수정 된 행의 수 :', cursor.rowcount)

            conn.commit()

except mysql.connector.Error as err:
    print("DB 에러:", err)
```
